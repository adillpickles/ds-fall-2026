### Clean the Data

Well, we can’t get our answers with what we currently have, so you’ll have to make some changes.

Let’s not worry about anything below the first 5 values for now. Convert the top 5 to a single canonical value―say, `'US'`, which is nice and short.

```python
# Find the five most common country labels in the USD tech data.
top_five_countries = df_salary_tech_usd['country'].value_counts().head(5).index

# Make a condition for rows using one of those labels.
top_five_condition = df_salary_tech_usd['country'].isin(top_five_countries)

# Give those common U.S. labels one short common label.
df_salary_tech_usd.loc[top_five_condition, 'country'] = 'US'
```

Have a look at the count of each unique country again now.

```python
# Count countries again after standardizing the top five labels.
df_salary_tech_usd['country'].value_counts()
```

Did you notice anything interesting?

```python
# Remove extra spaces before checking the country text.
country_text = df_salary_tech_usd['country'].fillna('').str.strip()

# Look for common U.S. words anywhere in the country answer.
us_pattern = r'\b(us|usa|u\.s\.a?\.?|united states?|america)\b'

# Make a True or False mask for likely U.S. answers.
is_us = country_text.str.contains(us_pattern, case=False, regex=True, na=False)

# Change those likely U.S. answers to the same US label.
df_salary_tech_usd.loc[is_us, 'country'] = 'US'
```

```python
# Count the country labels after the bonus cleanup.
df_salary_tech_usd['country'].value_counts()
```

It’s looking good so far. Let’s find out the minimum, mean, and maximum (in that order) salary by state, sorted by the mean in descending order.

```python
# Keep the cleaned US records so the state comparison is meaningful.
df_salary_us = df_salary_tech_usd[df_salary_tech_usd['country'] == 'US'].copy()

# The salary column is still text, so a mean cannot be calculated yet.
# df_salary_us.groupby('state')['salary'].agg(['min', 'mean', 'max'])

# Look at a few state and salary values before fixing the type.
df_salary_us[['state', 'salary']].head()
```

Well, pooh! We forgot that `salary` isn’t numeric. Something wrong must be fixed.

```python
# Remove commas so salary strings can be read as numbers.
df_salary_us['salary'] = df_salary_us['salary'].str.replace(',', '', regex=False)

# Convert the cleaned salary text into numbers.
df_salary_us['salary'] = pd.to_numeric(df_salary_us['salary'], errors='coerce')

# Remove rows with no usable salary or state.
df_salary_us = df_salary_us.dropna(subset=['salary', 'state']).copy()

# Check the salary type after the fix.
print(df_salary_us['salary'].dtype)
```

Let’s try that again.

```python
# Group the salaries by state and calculate the three requested values.
salary_by_state = df_salary_us.groupby('state')['salary'].agg(['min', 'mean', 'max'])

# Put the state with the largest average salary first.
salary_by_state = salary_by_state.sort_values('mean', ascending=False)

# Display the finished table.
salary_by_state
```

That did the trick! Now let’s narrow this to data 2021 and 2022 just because (lel). *(Hint: that timestamp column may not be a temporal type right now.)*

```python
# Turn the survey timestamp text into datetime values.
df_salary_us['timestamp'] = pd.to_datetime(df_salary_us['timestamp'], errors='coerce')

# Get the year from each timestamp.
df_salary_us['year'] = df_salary_us['timestamp'].dt.year

# Keep responses from 2021, 2022, or 2023.
recent_condition = df_salary_us['year'].isin([2021, 2022, 2023])

# Save the filtered rows in a new DataFrame.
df_salary_recent = df_salary_us[recent_condition].copy()

# Calculate the summary again for the recent rows.
recent_salary_by_state = df_salary_recent.groupby('state')['salary'].agg(['min', 'mean', 'max'])

# Sort by the average salary from highest to lowest.
recent_salary_by_state = recent_salary_by_state.sort_values('mean', ascending=False)

# Display the result.
recent_salary_by_state
```

### Bonus #0

Apply the same steps as before to `df_salary`, but do not filter for any specific industry. Do perform the other data cleaning stuff, and get to a point where you can generate the minimum, mean, and maximum by state.

```python
# Start with the renamed salary data and keep every industry.
df_salary_all = df_salary.copy()

# Keep only salaries reported in U.S. dollars.
all_usd_condition = df_salary_all['currency'] == 'USD'
df_salary_all_usd = df_salary_all[all_usd_condition].copy()

# Remove extra spaces before checking the country text.
all_country_text = df_salary_all_usd['country'].fillna('').str.strip()

# Look for the same common U.S. labels used above.
all_is_us = all_country_text.str.contains(us_pattern, case=False, regex=True, na=False)

# Standardize those country labels.
df_salary_all_usd.loc[all_is_us, 'country'] = 'US'

# Keep the U.S. records without filtering by industry.
df_salary_all_us = df_salary_all_usd[df_salary_all_usd['country'] == 'US'].copy()

# Remove commas from the salary text.
df_salary_all_us['salary'] = df_salary_all_us['salary'].str.replace(',', '', regex=False)

# Convert salary text to numbers.
df_salary_all_us['salary'] = pd.to_numeric(df_salary_all_us['salary'], errors='coerce')

# Remove rows missing either salary or state.
df_salary_all_us = df_salary_all_us.dropna(subset=['salary', 'state']).copy()

# Calculate minimum, mean, and maximum salary by state.
summary_by_state = df_salary_all_us.groupby('state')['salary'].agg(['min', 'mean', 'max'])

# Sort the states by their mean salary.
summary_by_state = summary_by_state.sort_values('mean', ascending=False)

# Display the summary for all industries.
summary_by_state
```

### Bonus #2

Filter out the non-single-states (e.g., `'California, Colorado'`) in the most elegant way possible (i.e., *not* by blacklisting all the bad values).

```python
# Mark rows whose state answer does not contain a comma.
single_state_condition = ~df_salary_all_us['state'].str.contains(',', na=False)

# Also require a state value to be present.
single_state_condition = single_state_condition & df_salary_all_us['state'].notna()

# Keep only answers with one state.
df_salary_single_state = df_salary_all_us[single_state_condition].copy()

# Check the remaining state counts.
df_salary_single_state['state'].value_counts()
```

### Bonus #3

Show the quantiles instead of just minimum, mean, and maximum―say 0%, 5%, 25%, 50%, 75%, 95%, and 100%. Outliers may be deceiving.

Sort by whatever interests you―like say the *50th* percentile.

And throw in a count by state too. It would be interesting to know how many data points contribute to the figures for each state. (*Hint: your nice formatting from Bonus #1 might not work this time around.* 😜)

```python
# List the percentiles we want to see.
quantile_levels = [0.00, 0.05, 0.25, 0.50, 0.75, 0.95, 1.00]

# Calculate each percentile for each state.
state_quantiles = df_salary_single_state.groupby('state')['salary'].quantile(quantile_levels)

# Turn the percentile index into columns.
state_quantiles = state_quantiles.unstack()

# Give the percentile columns readable names.
state_quantiles.columns = ['0%', '5%', '25%', '50%', '75%', '95%', '100%']

# Count the number of salary records for each state.
state_counts = df_salary_single_state['state'].value_counts().rename('count')

# Join the counts to the percentile table using the state index.
state_summary = state_quantiles.join(state_counts)

# Sort by the median, which is the 50th percentile.
state_summary = state_summary.sort_values('50%', ascending=False)

# Display the final table.
state_summary
```
